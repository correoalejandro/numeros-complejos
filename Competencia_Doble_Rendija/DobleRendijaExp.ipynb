# Experimento: Competencia de la doble rendija



### Introducción
El experimento de la doble rendija, propuesto por Thomas Young en 1801, demuestra la naturaleza ondulatoria de la luz al mostrar un patrón de franjas brillantes y oscuras en la pantalla cuando un haz coherente atraviesa dos rendijas estrechas y cercanas. Cada rendija actúa como una fuente secundaria de ondas, que al superponerse producen regiones de interferencia constructiva (máximos) y destructiva (mínimos). En el contexto cuántico, este experimento adquiere mayor relevancia, pues al repetirse con electrones u otras partículas se observa el mismo patrón, lo que confirma la dualidad onda-partícula y plantea interrogantes sobre el papel de la observación en los fenómenos físicos.

En nuestra práctica reproducimos este montaje a pequeña escala utilizando materiales accesibles: un láser de baja potencia como fuente de luz coherente, cartón paja y papel aluminio para construir la barrera con dos rendijas, y una pantalla blanca para observar el patrón de interferencia.

## Objetivos

- Reproducir el experimento de la doble rendija utilizando un láser y materiales accesibles (cartón paja y papel aluminio).
- Observar y registrar el patrón de interferencia resultante, identificando franjas brillantes y oscuras.
- Medir experimentalmente la separación entre franjas y compararla con los valores teóricos.
- Complementar el montaje con una simulación computacional que modele la interferencia y la difracción, y contrastar ambos resultados.

## Simulaciones y marco teórico  

El experimento de la doble rendija combina dos fenómenos ondulatorios fundamentales: **interferencia** y **difracción**.  
Para entenderlos y compararlos con el montaje real, se desarrollaron varias simulaciones en Python, cada una basada en un modelo distinto.  



### Simulación 1: Interferencia pura de dos rendijas  
Modelo:  

    y_m = (m * λ * L) / d

- Describe la posición de las franjas brillantes en la pantalla.  
- Solo considera la superposición de ondas procedentes de dos rendijas puntuales.  
- Resultado: franjas igualmente espaciadas y con la misma intensidad.  



### Simulación 2: Interferencia modulada por difracción  
Modelo:  

    I(θ) ∝ cos²(γ) * (sin β / β)²
    γ = (π * d * sin θ) / λ
    β = (π * a * sin θ) / λ

- Incluye el **ancho de cada rendija (a)**, por lo que aparece una envolvente de difracción.  
- El patrón de interferencia se mantiene, pero la intensidad de las franjas decrece a medida que nos alejamos del centro.  
- Resultado: franjas centrales más intensas, laterales más débiles.  



### Simulación 3: Variación de la separación entre rendijas (d)  
Modelo:  

    y_m = (m * λ * L) / d

- Se usó el mismo modelo de interferencia pura, pero variando d.  
- Al aumentar d → las franjas se acercan.  
- Al disminuir d → las franjas se separan más.  
- Resultado: mostró cómo la geometría del montaje controla el espaciado del patrón.  



### Simulación 4: Variación del ancho de rendija (a)  
Modelo:  

    I_difracción(θ) ∝ (sin β / β)²
    β = (π * a * sin θ) / λ

- Se centró en cómo cambia la envolvente de difracción al variar a.  
- Rendijas más anchas → envolvente más estrecha → menos franjas visibles.  
- Rendijas más finas → envolvente más ancha → más franjas dentro de la máxima central.  
- Resultado: evidenció el rol de la difracción en la visibilidad y contraste del patrón.  



### Resultado escogido de la simulación

En este informe incluimos únicamente el código de la simulación más cercana a los resultados reales, con el fin de hacer una comparación directa con el experimento.

#### Anexo: Código de simulación mas cercana

```python
import numpy as np
import matplotlib.pyplot as plt
slit_distance = 0

slit_distance = 0.0002     # distancia entre rendijas (m)
slit_width = 1e-4          # ancho de cada rendija (m) -> AJUSTABLE
wavelength = 670e-9        # longitud de onda (m) ~ luz roja
screen_distance = 0.4      # distancia a la pantalla (m) 40 cm
screen_width = 0.1         # ancho de la pantalla (m) 10 cm
num_points = 5000          # puntos de cálculo en la pantalla
print(f"Distancia entre rendijas: {slit_distance}")


# --
# Paso 2: Puntos en la pantalla
# --
screen_points = np.linspace(-screen_width/2, screen_width/2, num_points)
theta = np.arctan(screen_points / screen_distance)

# --
# Paso 3: Interferencia
# --
path_diff = (slit_distance * screen_points) / screen_distance
phase_diff = (2 * np.pi / wavelength) * path_diff
amplitude1 = np.ones_like(screen_points)
amplitude2 = np.exp(1j * phase_diff)
resultant_amplitude = amplitude1 + amplitude2
interference_intensity = np.abs(resultant_amplitude)**2

# --
# Paso 4: Difracción (envolvente sinc²)
# --
beta = (np.pi * slit_width * np.sin(theta)) / wavelength
beta[beta == 0] = 1e-12
envelope = (np.sin(beta) / beta)**2

# Intensidad total
intensity = interference_intensity * envelope
intensity /= np.max(intensity)

# --
# Paso 5: Gráfico en 4 cuadrantes
# --
plt.figure(figsize=(10,6))

# Patrón normal (arriba)
plt.plot(screen_points*1000, intensity, color='red', lw=0.8, label='Intensidad +')

# Patrón reflejado (abajo)
plt.plot(screen_points*1000, -intensity, color='red', lw=0.8, label='Intensidad -')

plt.xlim(-5, 5)  # en mm
plt.ylim(-1.1, 1.1)

plt.axhline(0, color='black', lw=0.8)  # eje horizontal
plt.axvline(0, color='black', lw=0.8)  # eje vertical

plt.xlabel('Posición en la pantalla (mm)')
plt.ylabel('Intensidad normalizada (±)')
plt.title('Patrón de doble rendija con difracción (4 cuadrantes)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## Comparación entre experimento y simulación  

La comparación entre el patrón obtenido en el montaje experimental y el calculado con el modelo completo de interferencia + difracción muestra una concordancia importante tanto en la forma general como en los valores numéricos.

### Coincidencias principales  
- **Franja central:** en ambos casos la franja central es la más intensa.  
- **Franjas laterales:** las dos primeras franjas a cada lado tienen una intensidad cercana a la central; a partir de ahí su intensidad cae rápidamente.  
- **Separación de franjas:** experimentalmente se midió una separación promedio de **≈ 1,5 mm**, mientras que la simulación predice un valor muy cercano (**≈ 1,6 mm**).  
- **Forma general:** en la simulación las franjas son perfectamente simétricas y paralelas, mientras que en el experimento aparecen distorsionadas (de | | a <>), formando visualmente una figura similar a `<0>`.  

### Diferencias observadas  
- **Contraste:** la simulación muestra franjas con contraste perfecto (máximos bien definidos y mínimos prácticamente nulos), mientras que en el experimento los mínimos no llegan a apagarse del todo.  
- **Regularidad:** en el modelo las franjas son uniformes y rectas; en el montaje real se observan inclinaciones y variaciones en la forma de las franjas por cortes irregulares y alineación imperfecta del láser.  
- **Ruido y dispersión:** la foto experimental presenta brillo difuso y ruido óptico, propios de la luz ambiente y la dispersión del aluminio.

#### Anexo:

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
wavelength = 670e-9   # longitud de onda (m)
L = 0.4               # distancia a la pantalla (m)

# Datos experimentales (mm)
franjas = np.array([1.57, 1.497, 1.481, 1.421])   # separaciones de franjas brillantes (mm)
delta_y_exp = franjas.mean() * 1e-3               # promedio a metros

# Estimación de separación entre rendijas (d)
d_est = wavelength * L / delta_y_exp
print(f"Separación estimada entre rendijas d = {d_est*1e3:.3f} mm")

# Simulación interferencia × difracción con d_est
a = 1e-4    # ancho de rendija (m) = 0,10 mm
d = d_est
x = np.linspace(-0.004, 0.004, 2000)  # posiciones en la pantalla (±4 mm)
theta = np.arctan(x / L)

beta = (np.pi * a * np.sin(theta)) / wavelength
gamma = (np.pi * d * np.sin(theta)) / wavelength

sinc2 = np.ones_like(beta)
mask = np.abs(beta) > 1e-12
sinc2[mask] = (np.sin(beta[mask]) / beta[mask])**2

I_sim = sinc2 * (np.cos(gamma)**2)
I_sim /= I_sim.max()

# Máximos teóricos a partir de la separación experimental
x_max_theory_mm = np.arange(-5, 6) * (delta_y_exp * 1e3)  # en mm

# Gráfico
plt.figure(figsize=(8,4))
plt.plot(x*1e3, I_sim, '-', label='Simulación (normalizada)')
plt.scatter(x_max_theory_mm, np.ones_like(x_max_theory_mm)*0.95,
            color='red', marker='v', label='Máximos teóricos (exp)')
plt.axvline(0, color='k', lw=0.6)
plt.xlim(-4, 4)
plt.xlabel("Posición en la pantalla (mm)")
plt.ylabel("Intensidad normalizada")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Discusión de errores y limitaciones  

Durante el montaje en el laboratorio de óptica se presentaron varias limitaciones que afectaron la calidad del patrón observado:  

- **Precisión de las rendijas:** al ser realizadas manualmente con cúter sobre cartón paja y papel aluminio, las rendijas no quedaron perfectamente rectas ni uniformes. Esto generó la distorsión de las franjas laterales, que pasaron de verse casi verticales (| |) a inclinarse en forma de ángulo (<>), dando la apariencia de un símbolo <0>.  
- **Alineación del láser:** pequeñas desviaciones respecto a la perpendicularidad de la barrera afectaron la simetría del patrón.  
- **Saturación de la franja central:** la máxima intensidad provocó sobreexposición en la cámara, dificultando medir con precisión la relación de intensidades con las franjas laterales.  
- **Registro fotográfico:** aunque la iluminación del laboratorio fue adecuada, el método de captura con cámara de celular introdujo ruido y limitó la fidelidad de los detalles en las franjas más débiles.
